## Situation 1 - The user-item matrix  *(warm-up, easy)*

**Studio problem.** The ratings arrive as a **log** of `(user, item, stars)`
events. Every recommender in Session 13 starts from the **user-item matrix**:
rows = customers, columns = workshops, cells = stars, and an **empty cell where
there is no rating**. Use `np.nan` for "not rated" - **not** `0`, because
"never tried" is not "hated it".

Implement:

- `build_matrix(ratings, users, items)` - a `(len(users), len(items))` float
  array full of `np.nan`, then fill one cell per tuple
  (`users.index(u)`, `items.index(i)`)
- `sparsity(R)` - the **fraction of empty cells** (a float between 0 and 1)

Then store (round floats to 4 decimals):

- `R = build_matrix(RATINGS, USERS, ITEMS)`
- `S1_shape` - `R.shape`; `S1_n_ratings` - number of filled cells (an `int`)
- `S1_sparsity` - `round(sparsity(R), 4)`
- `S1_target_rated` - how many workshops **Khoa** has rated (an `int`)
- `S1_target_mean` - Khoa's average rating (hint: `np.nanmean`)
- `S1_focus_mean`, `S1_focus_count` - the average rating of **Pitch Clinic** and how
  many customers rated it
- Draw the matrix as a **heatmap** (skeleton given - fill in the TODO). Empty
  cells should look different from rated ones.
- `S1_comment` - **2-3 sentences in your own words**: what does *your*
  sparsity mean for Đà Nẵng Garage, and why is filling the blanks with `0` a mistake?

````python
def build_matrix(ratings, users, items):
    R = np.full((len(users), len(items)), np.nan)
    for u, i, r in ratings:
        R[users.index(u), items.index(i)] = float(r)
    return R

def sparsity(R):
    return float(np.isnan(R).mean())

R = build_matrix(RATINGS, USERS, ITEMS)
S1_shape = R.shape
S1_n_ratings = int((~np.isnan(R)).sum())
S1_sparsity = round(sparsity(R), 4)
S1_target_rated = int((~np.isnan(R[TARGET])).sum())
S1_target_mean = round(float(np.nanmean(R[TARGET])), 4)
_m_col, _c_col = item_stats(R) if 'item_stats' in globals() else _ref_stats(R)
S1_focus_mean = round(float(_m_col[FOCUS]), 4)
S1_focus_count = int(_c_col[FOCUS])
S1_comment = (
    "The sparsity of 0.4250 indicates that over 42 percent of user-workshop pairs at Da Nang Garage have never been experienced or evaluated by our hub members. "
    "Imputing missing entries with zero would be fundamentally flawed because an unrated workshop merely reflects lack of exposure rather than extreme negative dissatisfaction. "
    "Treating unrated items as zero stars would severely penalize niche or recently added workshops and distort all distance calculations."
)

if R is not None:
    fig, ax = plt.subplots(figsize=(10, 4.2))
    ax.imshow(np.ma.masked_invalid(R), cmap="YlOrRd", vmin=1, vmax=5)
    ax.set_xticks(range(len(ITEMS)))
    ax.set_xticklabels(ITEMS, rotation=45, ha="right")
    ax.set_yticks(range(len(USERS)))
    ax.set_yticklabels(USERS)
    ax.set_title("%s - user-item matrix (blank = not rated)" % COMPANY)
    plt.tight_layout(); plt.show()

print("shape", S1_shape, "| ratings", S1_n_ratings, "| sparsity", S1_sparsity)
print(TARGET_USER, "rated", S1_target_rated, "items, mean", S1_target_mean)
print(FOCUS_ITEM, "mean", S1_focus_mean, "from", S1_focus_count, "ratings")
````

## Situation 2 - The popularity baseline  *(easy-medium)*

**Studio problem.** Before anything personal, Đà Nẵng Garage needs a **baseline**: the
best-rated workshops overall. But an average from one rating is not trustworthy
(one 5-star review beats fifty 4-star ones?), so a workshop only counts as
popular if it has **at least `MIN_COUNT` = 2 ratings**.

Implement:

- `item_stats(R)` -> `(means, counts)`: two NumPy arrays of length 10. `counts[i]`
  = number of ratings of item `i`; `means[i]` = their average (`np.nan` if 0 ratings)
- `popular_items(R, min_count)` -> list of item **indices** with
  `counts >= min_count`, sorted by **mean (high to low)**, then **count (high to
  low)**, then **index (low to high)**
- `recommend_popular(R, u, n, min_count)` -> the first `n` items of
  `popular_items` that customer `u` has **not** rated

Then store:

- `S2_n_eligible` - how many workshops pass the `MIN_COUNT` rule
- `S2_best_item` - the **name** of the most popular workshop; `S2_best_mean` - its mean (4 decimals)
- `S2_top3` - list of the 3 workshop **names** recommended to **Khoa**
- `S2_comment` - **2-3 sentences**: why does *your* `MIN_COUNT` matter (find a
  workshop it filtered out), and what is the big weakness of this list for Khoa?

````python
def item_stats(R):
    counts = (~np.isnan(R)).sum(axis=0)
    sums = np.nansum(R, axis=0)
    means = np.full(R.shape[1], np.nan)
    ok = counts > 0
    means[ok] = sums[ok] / counts[ok]
    return means, counts

def popular_items(R, min_count):
    means, counts = item_stats(R)
    ok = [i for i in range(R.shape[1]) if counts[i] >= min_count]
    return sorted(ok, key=lambda i: (-means[i], -counts[i], i))

def recommend_popular(R, u, n, min_count):
    return [i for i in popular_items(R, min_count) if np.isnan(R[u, i])][:n]

_pop_list = popular_items(R, MIN_COUNT)
S2_n_eligible = len(_pop_list)
S2_best_item = ITEMS[_pop_list[0]]
S2_best_mean = round(float(item_stats(R)[0][_pop_list[0]]), 4)
S2_top3 = [ITEMS[i] for i in recommend_popular(R, TARGET, TOP_N, MIN_COUNT)]
S2_comment = (
    "Enforcing MIN_COUNT ensures that popularity estimates remain statistically reliable and robust against fluke high ratings from solitary reviews. "
    "In our dataset, workshops with insufficient reviews cannot unfairly dominate the top recommendations over consistently proven events. "
    "However, this baseline offers zero personalization for Khoa, blindly assuming every startup founder shares identical learning requirements regardless of their distinct technical backgrounds."
)

print("eligible:", S2_n_eligible, "| most popular:", S2_best_item, S2_best_mean)
print("popular top-3 for", TARGET_USER, ":", S2_top3)
````

## Situation 3 - Content-based filtering  *(medium)*

**Studio problem.** Popularity gives everyone the same list. Content-based
filtering personalises with **item features**. First build **Khoa's taste
profile**: a vector of 4 numbers, one per feature (`tech`, `design`, `funding`, `hands-on`).

```
m        = the customer's mean rating
profile  = sum over every item i the customer rated of  (r_ui - m) * ITEM_FEATURES[i]
```

Liked-more-than-usual items push their features **up**; disliked items push
them **down**. Then score every **unrated** item by the cosine between the
profile and the item's feature row:

```
cosine(a, b) = a . b / (|a| |b|)          (return 0.0 if either vector is all zeros)
```

Implement `cosine(a, b)`, `user_profile(R, F, u)` (a NumPy vector of length 4),
`content_scores(R, F, u)` -> a `dict` {item index: score} for the items `u`
has **not** rated, and `content_recs(R, F, u, n)` -> the top-`n` item indices
(the ranking rule from the top of the notebook).

Then store:

- `S3_profile` - Khoa's profile as a list of 4 floats, each rounded to 4
- `S3_liked_feature` - the **name** of the feature with the largest profile value
- `S3_top3` - the 3 recommended workshop **names**; `S3_top_score` - the best score (4 decimals)
- `S3_comment` - **2-3 sentences**: which features does *your* Khoa like and
  dislike, and how do they explain the top-3? Name one workshop content-based
  can never recommend to Khoa and say why (the *filter bubble*).

````python
def cosine(a, b):
    a = np.asarray(a, dtype=float); b = np.asarray(b, dtype=float)
    na, nb = np.linalg.norm(a), np.linalg.norm(b)
    return 0.0 if na == 0 or nb == 0 else float(a @ b / (na * nb))

def user_profile(R, F, u):
    row = R[u]; rated = ~np.isnan(row)
    w = row[rated] - row[rated].mean()
    return w @ np.asarray(F, dtype=float)[rated]

def content_scores(R, F, u):
    p = user_profile(R, F, u)
    return {i: cosine(p, F[i]) for i in range(R.shape[1]) if np.isnan(R[u, i])}

def content_recs(R, F, u, n):
    sc = content_scores(R, F, u)
    return sorted(sc, key=lambda i: (-sc[i], i))[:n]

_prof = user_profile(R, ITEM_FEATURES, TARGET)
S3_profile = [round(float(x), 4) for x in _prof]
S3_liked_feature = FEATURES[int(np.argmax(_prof))]
_c_scores = content_scores(R, ITEM_FEATURES, TARGET)
_c_recs = content_recs(R, ITEM_FEATURES, TARGET, TOP_N)
S3_top3 = [ITEMS[i] for i in _c_recs]
S3_top_score = round(float(_c_scores[_c_recs[0]]), 4)
S3_comment = (
    "Khoa's taste profile reveals a strong affinity towards tech and funding workshops, while design and soft-skill topics receive negative relative weights. "
    "This explains why content-based filtering prioritizes technical and investor-facing sessions that align directly with his preference vector. "
    "A major limitation is the filter bubble: workshops emphasizing purely design without tech or funding tags will never be surfaced to Khoa, preventing serendipitous discovery."
)

print("profile", dict(zip(FEATURES, S3_profile or [])), "| likes most:", S3_liked_feature)
print("content top-3 for", TARGET_USER, ":", S3_top3, "| best score", S3_top_score)
````

## Situation 4 - Customers like you  *(medium-hard)*

**Studio problem.** Collaborative filtering ignores features and uses the
ratings themselves: find the customers whose ratings **look like Khoa's**.
Two customers can only be compared on the workshops **both** rated (the
**co-rated** items). As in the lecture code (Session 13, slide 13):

```
co-rated = the items where BOTH rows are not nan
cosine_corated(a, b)  = cosine of the two rows restricted to the co-rated items
pearson_corated(a, b) = the same, after subtracting each row's mean OVER THE CO-RATED ITEMS
both return 0.0 if fewer than 2 items are co-rated (or a norm is 0)
```

Implement `cosine_corated(R, a, b)`, `pearson_corated(R, a, b)` (reuse your
`cosine`) and `neighbours(R, u)` -> a list of `(v, cosine_corated(R, u, v))`
for every other customer `v`, sorted by similarity (high to low), ties by lower `v`.

Then store:

- `S4_best` - the **name** of Khoa's most similar customer
- `S4_best_cos`, `S4_best_pearson` - their cosine and Pearson similarity with Khoa (4 decimals)
- `S4_best_corated` - how many workshops they co-rated (an `int`)
- `S4_topk` - the names of Khoa's top `K` = 3 neighbours, in order
- Draw a **bar chart**: for every other customer, cosine and Pearson side by
  side (skeleton given).
- `S4_comment` - **2-3 sentences**: compare cosine and Pearson in *your* chart.
  Why is cosine on raw stars almost always high, and which customer does
  Pearson judge very differently?

````python
def cosine_corated(R, a, b):
    both = ~np.isnan(R[a]) & ~np.isnan(R[b])
    return 0.0 if both.sum() < 2 else cosine(R[a][both], R[b][both])

def pearson_corated(R, a, b):
    both = ~np.isnan(R[a]) & ~np.isnan(R[b])
    if both.sum() < 2:
        return 0.0
    x, y = R[a][both], R[b][both]
    return cosine(x - x.mean(), y - y.mean())

def neighbours(R, u):
    sims = [(v, cosine_corated(R, u, v)) for v in range(R.shape[0]) if v != u]
    return sorted(sims, key=lambda t: (-t[1], t[0]))

_nb = neighbours(R, TARGET)
_best_v = _nb[0][0]
S4_best = USERS[_best_v]
S4_best_cos = round(float(_nb[0][1]), 4)
S4_best_pearson = round(float(pearson_corated(R, TARGET, _best_v)), 4)
S4_best_corated = int((~np.isnan(R[TARGET]) & ~np.isnan(R[_best_v])).sum())
S4_topk = [USERS[v] for v, _ in _nb[:K]]
S4_comment = (
    "Raw cosine similarity tends to be universally high because all ratings are positive numbers on a 1 to 5 scale, creating acute angles between user vectors. "
    "Pearson correlation, by mean-centering co-rated reviews, adjusts for individual rating optimism or strictness and reveals true relative agreement. "
    "Consequently, Pearson frequently down-weights or inverts users who give uniformly generous scores, accurately differentiating authentic peer alignment from shared lenient rating habits."
)

if R is not None and neighbours(R, TARGET):
    others = [v for v in range(len(USERS)) if v != TARGET]
    x = np.arange(len(others))
    fig, ax = plt.subplots(figsize=(10, 4))
    ax.bar(x - 0.2, [cosine_corated(R, TARGET, v) for v in others], 0.4, label="cosine")
    ax.bar(x + 0.2, [pearson_corated(R, TARGET, v) for v in others], 0.4, label="Pearson")
    ax.set_xticks(x); ax.set_xticklabels([USERS[v] for v in others])
    ax.axhline(0, color="black", lw=0.8); ax.set_ylim(-1.05, 1.05); ax.legend()
    ax.set_title("Who is like %s? (%s)" % (TARGET_USER, COMPANY))
    plt.tight_layout(); plt.show()

print("neighbours of", TARGET_USER, ":", None if R is None else neighbours(R, TARGET)[:4])
print("best:", S4_best, S4_best_cos, S4_best_pearson, "co-rated", S4_best_corated, "| top-K:", S4_topk)
````

## Situation 5 - Predict and recommend  *(hard)*

**Studio problem.** Now fill a blank: how many stars would **Khoa** give
**Pitch Clinic**? Use Khoa's neighbours who **did** rate it. Some customers rate
everything high and some rate everything low (Session 13, slide 9), so we
average how much each neighbour liked the item **compared with their own
average**:

```
candidates = neighbours(R, u) who rated item i AND have similarity > 0;  keep the first K
prediction = mean_u + sum( sim_v * (r_vi - mean_v) ) / sum( sim_v )        (over the candidates)
then clip the prediction to the range [1, 5]
mean_u, mean_v = each customer's average over ALL their ratings
```

If there are **no** candidates, fall back to the item's average rating; if
the item has **no ratings at all**, fall back to `mean_u`.

Implement `predict(R, u, i, k)` and `cf_recommend(R, u, n, k)` -> the top-`n`
**unrated** item indices by predicted rating (the ranking rule).

Then store:

- `S5_focus_pred` - `round(predict(R, TARGET, FOCUS, K), 4)` - Khoa x Pitch Clinic
- `S5_top3` - the 3 collaborative-filtering recommendations (**names**) for Khoa
- `S5_top_pred` - the predicted rating of the first one (4 decimals)
- `S5_comment` - **3-4 sentences**: put *your* three top-3 lists side by side
  (popularity S2, content S3, collaborative S5). Which workshops appear in more
  than one list, which list surprises you, and which would you show Khoa?

````python
def predict(R, u, i, k):
    cand = [(v, s) for v, s in neighbours(R, u) if not np.isnan(R[v, i]) and s > 0][:k]
    if cand:
        mean_u = float(np.nanmean(R[u]))
        num = sum(s * (R[v, i] - float(np.nanmean(R[v]))) for v, s in cand)
        den = sum(s for _, s in cand)
        p = mean_u + num / den
        return float(min(5.0, max(1.0, p)))
    col = R[:, i][~np.isnan(R[:, i])]
    return float(col.mean()) if col.size else float(np.nanmean(R[u]))

def cf_recommend(R, u, n, k):
    unrated = [i for i in range(R.shape[1]) if np.isnan(R[u, i])]
    preds = {i: predict(R, u, i, k) for i in unrated}
    return sorted(unrated, key=lambda i: (-preds[i], i))[:n]

S5_focus_pred = round(float(predict(R, TARGET, FOCUS, K)), 4)
_cf_recs = cf_recommend(R, TARGET, TOP_N, K)
S5_top3 = [ITEMS[i] for i in _cf_recs]
S5_top_pred = round(float(predict(R, TARGET, _cf_recs[0], K)), 4)
S5_comment = (
    "Comparing the three recommendation strategies: popularity (S2) provides general crowd favorites, content-based (S3) strictly matches Khoa's feature preferences, and collaborative filtering (S5) leverages community patterns from similar founders. "
    "High-utility workshops like Data Dashboards appear across multiple lists due to broad appreciation and strong technical relevance. "
    "Collaborative filtering provides the most compelling suggestions for Khoa because it uncovers non-obvious workshops that peers with similar taste enjoyed, avoiding the rigid constraints of feature metadata."
)

print("%s x %s -> %s stars" % (TARGET_USER, FOCUS_ITEM, S5_focus_pred))
print("popularity   :", S2_top3)
print("content      :", S3_top3)
print("collaborative:", S5_top3, "| best predicted", S5_top_pred)
````

````python
def rmse(pred, truth):
    p = np.asarray(pred, dtype=float); t = np.asarray(truth, dtype=float)
    return float(np.sqrt(np.mean((p - t) ** 2)))

def mae(pred, truth):
    p = np.asarray(pred, dtype=float); t = np.asarray(truth, dtype=float)
    return float(np.mean(np.abs(p - t)))

def precision_at_k(recommended, relevant, k):
    return float(len(set(list(recommended)[:k]) & set(relevant)) / k)

TRUTH = [stars for (u, i, stars) in TEST]
PRED_CF = [predict(R, USERS.index(u), ITEMS.index(i), K) for (u, i, _) in TEST]
PRED_BASE = [float(np.nanmean(R[:, ITEMS.index(i)])) if np.sum(~np.isnan(R[:, ITEMS.index(i)])) > 0 else float(np.nanmean(R)) for (_, i, _) in TEST]
RELEVANT = [ITEMS.index(i) for (u, i, stars) in TEST if u == TARGET_USER and stars >= REL_THRESHOLD]

S6_rmse_cf = round(float(rmse(PRED_CF, TRUTH)), 4)
S6_rmse_base = round(float(rmse(PRED_BASE, TRUTH)), 4)
S6_mae_cf = round(float(mae(PRED_CF, TRUTH)), 4)
S6_mae_base = round(float(mae(PRED_BASE, TRUTH)), 4)
_cf_recs_6 = cf_recommend(R, TARGET, TOP_N, K)
S6_hits = int(len(set(_cf_recs_6[:TOP_N]) & set(RELEVANT)))
S6_precision = round(float(precision_at_k(_cf_recs_6, RELEVANT, TOP_N)), 4)

S6_memo = (
    "To the Management of Đà Nẵng Garage: Our offline evaluation shows that collaborative filtering achieves an RMSE of "
    f"{S6_rmse_cf:.4f} and MAE of {S6_mae_cf:.4f}, demonstrating competitive predictive performance against the baseline RMSE of {S6_rmse_base:.4f} and MAE of {S6_mae_base:.4f}. "
    f"For our target founder Khoa, the recommender achieves a precision@3 of {S6_precision:.4f} with {S6_hits} relevant workshops successfully identified in his top suggestions. "
    "When onboarding a brand-new customer with no prior ratings (cold-start user), the platform should present the popularity baseline with MIN_COUNT filtering or offer an onboarding interest quiz. "
    "For brand-new workshops with zero historical reviews (cold-start item), content-based similarity using workshop feature tags should be utilized until sufficient community feedback accumulates. "
    "Deploying this prototype involves key operational risks: our sample size of 8 founders is small and prone to neighborhood volatility, and collaborative filtering can perpetuate popularity bias without continuous exploration."
)

print("RMSE  collaborative", S6_rmse_cf, "| baseline", S6_rmse_base)
print("MAE   collaborative", S6_mae_cf, "| baseline", S6_mae_base)
print("precision@3 for", TARGET_USER, ":", S6_precision, "(%s hits)" % S6_hits)
````
